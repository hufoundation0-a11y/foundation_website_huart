```python
class taxi_cost:

    raise_perc = 0.2
    
    def __init__(self, cost, name):
        self.cost = cost
        self.name = name
        self.index_name = str(cost) + name + "__"

    def cal_cost(self):
        return self.cost * 100 if self.cost > 10 else 0 
          
    def cost_raise(self):
        self.cost *= 1 + taxi_cost.raise_perc
        return self.cost

    @classmethod
    def set_raise(cls, model_name):
        cost, name = model_name.split()
        return cls(float(cost), name)
        
        

taxi_1 = taxi_cost.set_raise("30 acura")
taxi_1.cost_raise()
taxi_1.cost
```
